#import libraries
import pandas as pd
import numpy as np


```python
# Create row data
file = {
    "Name":["Sahil","Rahul","Aman","priya","Naha"],
    "Age":[20,21,19,20,22],
    "Maths":[85,78,92,88,75],
    "Department":["Data Science","Computer Science","Data Science","Computer Science","Data Science"]
}
```

```python
student_makes = pd.DataFrame(file)
print(student_makes)
```

```python
print("First 5 rows :\n",student_makes.head())
print("Data type :\n",student_makes.dtypes)
print("\nshape :",student_makes.shape)
```

```python
numeric_data =student_makes[['Age','Maths']].values
print(numeric_data)
```

```python
print("shape : ",numeric_data.shape)
print("size : ",numeric_data.size)
print("dimension : ",numeric_data.ndim)
```

```python
A = np.array([[1,2],
              [3,4]])
B = np.array([[1,12],
              [8,4]])
print("Matrix A:\n",A,"\nMatrix B:\n",B)
```

```python
print("Addition:\n",A+B)
print("substraction :\n",A-B)
print("Multiplication :\n",A*B)
print("Dot product :\n",np.dot(A,B))
print('Transpose of A :\n',A.T)
print("Inverse of A :\n",np.linalg.inv(A))
print('Determinant of A: \n ',np.linalg.det(A))

#convert string data into numerical data
encoded_student_marks_department =pd.get_dummies(
    student_makes,
    columns=["Department"],
    dtype=int
)
print("\nOne-Hot Encoded Data:")
print(encoded_student_marks_department)
```
